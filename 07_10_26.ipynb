```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    # 1. Insert at Beginning
    def insert_begin(self, data):
        nn = Node(data)

        if self.head is None:
            self.head = nn
            return

        nn.next = self.head
        self.head = nn

    # 2. Insert at End
    def insert_end(self, data):
        nn = Node(data)

        if self.head is None:
            self.head = nn
            return

        temp = self.head

        while temp.next is not None:
            temp = temp.next

        temp.next = nn

    # 3. Insert at Any Position
    def insert_at_any_pos(self, pos, data):
        if pos <= 0:
            print("Invalid position")
            return

        if pos == 1:
            self.insert_begin(data)
            return

        temp = self.head
        i = 1

        while i < pos - 1 and temp is not None:
            temp = temp.next
            i += 1

        if temp is None:
            print("Insufficient Elements")
            return

        nn = Node(data)
        nn.next = temp.next
        temp.next = nn

    # 4. Count
    def count(self):
        temp = self.head
        c = 0

        while temp is not None:
            c += 1
            temp = temp.next

        return c

    # 5. Insert at Middle
    def add_at_mid(self, data):
        n = self.count()

        if n == 0:
            self.insert_begin(data)
            return

        mid = (n // 2) + 1
        self.insert_at_any_pos(mid, data)

    # 6. Delete at Beginning
    def delete_begin(self):
        if self.head is None:
            print("No Elements in the list")
            return

        self.head = self.head.next

    # 7. Delete at End
    def delete_end(self):
        if self.head is None:
            print("No Elements in the list to delete")
            return

        if self.head.next is None:
            self.head = None
            return

        temp = self.head
        prev = None

        while temp.next is not None:
            prev = temp
            temp = temp.next

        prev.next = None

    # 8. Delete at Any Position
    def delete_any_pos(self, pos):
        if pos <= 0:
            print("Invalid Position")
            return

        if pos == 1:
            self.delete_begin()
            return

        temp = self.head
        prev = None
        i = 1

        while i < pos and temp is not None:
            prev = temp
            temp = temp.next
            i += 1

        if temp is None:
            print("Insufficient elements")
            return

        prev.next = temp.next

    # 9. Delete at Middle
    def delete_mid(self):
        n = self.count()

        if n == 0:
            print("No Elements in the list")
            return

        mid = (n // 2) + 1
        self.delete_any_pos(mid)

    # 10. Search
    def search(self, data):
        temp = self.head
        pos = 1

        while temp is not None:
            if temp.data == data:
                return pos

            temp = temp.next
            pos += 1

        return -1

    # 11. Display
    def display(self):
        if self.head is None:
            print("List is empty")
            return

        temp = self.head

        while temp is not None:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


# Main Program
ll = SinglyLinkedList()

while True:

    print("\n----- SINGLY LINKED LIST -----")
    print("1. Insert at Beginning")
    print("2. Insert at End")
    print("3. Insert at Any Position")
    print("4. Insert at Middle")
    print("5. Delete at Beginning")
    print("6. Delete at End")
    print("7. Delete at Any Position")
    print("8. Delete at Middle")
    print("9. Count")
    print("10. Search")
    print("11. Display")
    print("12. Exit")

    choice = int(input("Enter your choice: "))

    match choice:

        case 1:
            data = int(input("Enter data: "))
            ll.insert_begin(data)

        case 2:
            data = int(input("Enter data: "))
            ll.insert_end(data)

        case 3:
            pos = int(input("Enter position: "))
            data = int(input("Enter data: "))
            ll.insert_at_any_pos(pos, data)

        case 4:
            data = int(input("Enter data: "))
            ll.add_at_mid(data)

        case 5:
            ll.delete_begin()

        case 6:
            ll.delete_end()

        case 7:
            pos = int(input("Enter position: "))
            ll.delete_any_pos(pos)

        case 8:
            ll.delete_mid()

        case 9:
            print("Number of elements:", ll.count())

        case 10:
            data = int(input("Enter data to search: "))
            pos = ll.search(data)

            if pos == -1:
                print("Element not found")
            else:
                print("Element found at position:", pos)

        case 11:
            ll.display()

        case 12:
            print("Program terminated")
            break

        case _:
            print("Invalid choice")
```

```python
class Node:
    def __init__(self, sid, name, marks):
        self.sid = sid
        self.name = name
        self.marks = marks
        self.next = None


class StudentList:
    def __init__(self):
        self.head = None

    # 1. Register student at beginning
    def insert_begin(self, sid, name, marks):
        if self.search(sid)[0] != -1:
            print("Student ID already exists")
            return

        nn = Node(sid, name, marks)

        nn.next = self.head
        self.head = nn

    # 2. Register student at end
    def insert_end(self, sid, name, marks):
        if self.search(sid)[0] != -1:
            print("Student ID already exists")
            return

        nn = Node(sid, name, marks)

        if self.head is None:
            self.head = nn
            return

        temp = self.head

        while temp.next is not None:
            temp = temp.next

        temp.next = nn

    # 3. Register student at specified position
    def insert_at_pos(self, pos, sid, name, marks):
        if pos <= 0:
            print("Invalid position")
            return

        if self.search(sid)[0] != -1:
            print("Student ID already exists")
            return

        if pos == 1:
            self.insert_begin(sid, name, marks)
            return

        temp = self.head
        i = 1

        while i < pos - 1 and temp is not None:
            temp = temp.next
            i += 1

        if temp is None:
            print("Invalid position")
            return

        nn = Node(sid, name, marks)
        nn.next = temp.next
        temp.next = nn

    # 4. Remove student using Student ID
    def delete_by_id(self, sid):
        if self.head is None:
            print("Student list is empty")
            return

        if self.head.sid == sid:
            self.head = self.head.next
            print("Student deleted")
            return

        temp = self.head

        while temp.next is not None:
            if temp.next.sid == sid:
                temp.next = temp.next.next
                print("Student deleted")
                return

            temp = temp.next

        print("Student ID not found")

    # 5. Search student using Student ID
    def search(self, sid):
        temp = self.head
        pos = 1

        while temp is not None:
            if temp.sid == sid:
                return pos, temp

            temp = temp.next
            pos += 1

        return -1, None

    # 6. Display all students
    def display(self):
        if self.head is None:
            print("Student list is empty")
            return

        temp = self.head

        while temp is not None:
            print("ID:", temp.sid,
                  "| Name:", temp.name,
                  "| Marks:", temp.marks)

            temp = temp.next

    # 7. Count students
    def count(self):
        temp = self.head
        c = 0

        while temp is not None:
            c += 1
            temp = temp.next

        return c

    # 8. Find student with highest marks
    def highest_marks(self):
        if self.head is None:
            print("Student list is empty")
            return

        temp = self.head
        highest = self.head

        while temp is not None:
            if temp.marks > highest.marks:
                highest = temp

            temp = temp.next

        print("Student with highest marks:")
        print("ID:", highest.sid)
        print("Name:", highest.name)
        print("Marks:", highest.marks)

    # 9. Reverse registration order
    def reverse(self):
        prev = None
        temp = self.head

        while temp is not None:
            next_node = temp.next
            temp.next = prev
            prev = temp
            temp = next_node

        self.head = prev

        print("Registration order reversed")


# Main Program
sl = StudentList()

while True:

    print("\n===== STUDENT RECORD MANAGER =====")
    print("1. Register student at beginning")
    print("2. Register student at end")
    print("3. Register student at position")
    print("4. Remove student using Student ID")
    print("5. Search student using Student ID")
    print("6. Display all students")
    print("7. Display total students")
    print("8. Find student with highest marks")
    print("9. Reverse registration order")
    print("10. Exit")

    choice = int(input("Enter your choice: "))

    match choice:

        case 1:
            sid = int(input("Enter Student ID: "))
            name = input("Enter Student Name: ")
            marks = float(input("Enter Marks: "))

            sl.insert_begin(sid, name, marks)

        case 2:
            sid = int(input("Enter Student ID: "))
            name = input("Enter Student Name: ")
            marks = float(input("Enter Marks: "))

            sl.insert_end(sid, name, marks)

        case 3:
            pos = int(input("Enter position: "))
            sid = int(input("Enter Student ID: "))
            name = input("Enter Student Name: ")
            marks = float(input("Enter Marks: "))

            sl.insert_at_pos(pos, sid, name, marks)

        case 4:
            sid = int(input("Enter Student ID to delete: "))

            sl.delete_by_id(sid)

        case 5:
            sid = int(input("Enter Student ID to search: "))

            pos, student = sl.search(sid)

            if pos == -1:
                print("Student ID not found")
            else:
                print("Student found")
                print("Position:", pos)
                print("ID:", student.sid)
                print("Name:", student.name)
                print("Marks:", student.marks)

        case 6:
            sl.display()

        case 7:
            print("Total students:", sl.count())

        case 8:
            sl.highest_marks()

        case 9:
            sl.reverse()

        case 10:
            print("Exiting application...")
            break

        case _:
            print("Invalid choice")1
```
