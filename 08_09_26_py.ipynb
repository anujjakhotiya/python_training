```python
import threading

mutex= threading.Lock()
balance=1000

def withdraw(amount):
  global balance
  mutex.acquire()

  if balance>=amount:
    balance-=amount
    print("Amount deducted",amount)
    print("Current balance",balance)
  else:
    print("Insufficient balance")

  mutex.release()

t1=threading.Thread(target=withdraw,args=(700,))
t2=threading.Thread(target=withdraw,args=(200,))

t1.start()
t2.start()

t1.join()
t2.join()

print("Final balance",balance)
```

```python
import threading

balance = 1000
flag = 0
def lock():
  global flag
  while (flag==1):
    pass
  flag=1
def unlock():
  global flag
  flag=0

def withdraw(amount):
    global balance, flag
    lock()
    if balance >= amount:
        balance -= amount
        print("Amount deducted:", amount)
        print("Remaining balance:", balance)
    else:
        print("Insufficient balance")
    unlock()

t1 = threading.Thread(target=withdraw, args=(700,))
t2 = threading.Thread(target=withdraw, args=(200,))

t1.start()
t2.start()

t1.join()
t2.join()

print("Final balance:", balance)
```
