PREFIX SUM


```python
l=list(map(int,input().split()))
l2=[]
sum=0
for i in range(len(l)):
  sum=sum+l[i]
  l2.append(sum)
print(l2)
```

SUFFIX SUM


```python
l=list(map(int,input().split()))
l2=[]
l.reverse()
sum=0
for i in range(len(l)):
  sum=sum+l[i]
  l2.append(sum)
print(l2)
```

```python
l=list(map(int,input().split()))
l2=[]
sum=0
for i in range(len(l)-1,-1,-1):
  sum=sum+l[i]
  l2.append(sum)
print(l2)
```
