```python
"""mobile:
name,brand,battery(private)
display()
charge()
use_phone(val)
get_battery()"""
class Mobile:
  __battery=100
  def __init__(self,name,brand):
    self.name=name
    self.brand=brand

  def display(self):
    print(f"Mobile Name: {self.name}, Brand: {self.brand}, Battery: {self.__battery}%")

  def charge(self, amount):
    if amount > 0:
      self.__battery = min(100, self.__battery + amount)
      print(f"Battery charged to {self.__battery}%")
    else:
      print("Charge amount must be positive.")

  def use_phone(self, usage_duration):
    if usage_duration > 0:
      self.__battery = max(0, self.__battery - usage_duration)
      print(f"Phone used for {usage_duration} units. Battery remaining: {self.__battery}%")
    else:
      print("Usage duration must be positive.")

  def get_battery(self):
    return self.__battery
```

```python
# Create a Mobile object
my_phone = Mobile("Pixel 6", "Google")

# Display initial state
print("Initial State:")
my_phone.display()

# Use the phone
my_phone.use_phone(20)
my_phone.display()

# Charge the phone
my_phone.charge(50)
my_phone.display()

# Use the phone again, checking battery getter
my_phone.use_phone(40)
print(f"Current battery level (using getter): {my_phone.get_battery()}%")
```

```python
from abc import ABC, abstractmethod
class Parent:
  @abstractmethod
  def degree(self,deg):
    pass
  @abstractmethod
  def job(self,j):
    pass

class Son(Parent):
  def degree(self,deg):
    print("I am going to learn:",deg)

  def job(self,j):
    print("I will be a:",j)

class Daughter(Parent):
  def degree(self,deg):
    print("I am going to learn:",deg)

  def job(self,j):
    print("I will be a:",j)

d=Daughter()
d.degree("MBBS")
d.job("Doctor")

s=Son()
s.degree("Engineering")
s.job("SDE")
```
