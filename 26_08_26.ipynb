Bank Account:

```python
class Bank:
  def __init__(self, name, account_no, initial_balance=0):
    self.name = name
    self.account_no = account_no
    self.balance = initial_balance

  def display(self):
    print("My name is:",self.name)
    print("My account number is:",self.account_no)
    print("My balance is:",self.balance)

  def deposit(self, amount):
    self.balance += amount
    print(f"Deposited {amount}. New balance: {self.balance}")

  def withdraw(self, amount):
    if self.balance >= amount:
      self.balance -= amount
      print(f"Withdrew {amount}. New balance: {self.balance}")
    else:
      print("Insufficient funds.")

  def check_balance(self):
    print("My balance is:",self.balance)

bank1 = Bank("Arya", 10, 100000)

print("Bank1 Details")
bank1.display()
bank1.deposit(2000)
bank1.withdraw(30000)
bank1.check_balance()
print()

bank2 = Bank("Bob", 20, 200000)

print("Bank2 Details")
bank2.display()
bank2.deposit(5000)
bank2.withdraw(10000)
bank2.check_balance()
print()
```

```python
# ==========================================
# 1. SINGLE INHERITANCE
# One parent class → One child class
# ==========================================

class Laptop:
    keyboard = 1
    mouse = 1

    def laptop_info(self):
        print("This is a laptop.")


class Lenovo(Laptop):          # Lenovo inherits from Laptop
    ram = 16
    ssd = 512

    def lenovo_info(self):
        print("This is a Lenovo laptop.")



l1 = Lenovo()

print("1. SINGLE INHERITANCE")
print(l1.keyboard)             # Takes variable from parent
print(l1.ram)                  # Takes variable from child

l1.laptop_info()               # Takes method from parent
l1.lenovo_info()               # Takes method from child


# ==========================================
# 2. MULTILEVEL INHERITANCE
# Grandparent → Parent → Child
# ==========================================

class Computer:
    monitor = "LCD"

    def computer_info(self):
        print("This is a computer.")


class Laptop2(Computer):       # Laptop2 inherits from Computer
    keyboard = 1

    def laptop_info(self):
        print("This is a laptop.")


class Lenovo2(Laptop2):        # Lenovo2 inherits from Laptop2
    ram = 16

    def lenovo_info(self):
        print("This is a Lenovo laptop.")


l2 = Lenovo2()

print("\n2. MULTILEVEL INHERITANCE")
print(l2.monitor)              # Takes variable from grandparent
print(l2.keyboard)             # Takes variable from parent
print(l2.ram)                  # Takes variable from child

l2.computer_info()             # Takes method from grandparent
l2.laptop_info()               # Takes method from parent
l2.lenovo_info()               # Takes method from child


# ==========================================
# 3. MULTIPLE INHERITANCE
# Multiple parent classes → One child class
# ==========================================

class Keyboard:
    keys = 104

    def keyboard_info(self):
        print("This has a keyboard.")


class Mouse:
    buttons = 3

    def mouse_info(self):
        print("This has a mouse.")


class Laptop3(Keyboard, Mouse):    # Inherits from two parent classes
    screen = 15

    def laptop_info(self):
        print("This is a laptop.")


l3 = Laptop3()

print("\n3. MULTIPLE INHERITANCE")
print(l3.keys)                  # Takes variable from Keyboard
print(l3.buttons)               # Takes variable from Mouse
print(l3.screen)                # Takes variable from child

l3.keyboard_info()              # Takes method from Keyboard
l3.mouse_info()                 # Takes method from Mouse
l3.laptop_info()                # Takes method from child


# ==========================================
# 4. HIERARCHICAL INHERITANCE
# One parent class → Multiple child classes
# ==========================================

class Laptop4:
    keyboard = 1
    mouse = 1

    def laptop_info(self):
        print("This is a laptop.")


class Lenovo4(Laptop4):         # Child 1 inherits from Laptop4
    ram = 16

    def lenovo_info(self):
        print("This is Lenovo.")


class Dell(Laptop4):            # Child 2 inherits from Laptop4
    ram = 8

    def dell_info(self):
        print("This is Dell.")


l4 = Lenovo4()

print("\n4. HIERARCHICAL INHERITANCE")
print(l4.keyboard)              # Takes variable from parent
print(l4.ram)                   # Takes variable from Lenovo

l4.laptop_info()                # Takes method from parent
l4.lenovo_info()                # Takes method from Lenovo


d = Dell()

print(d.mouse)                  # Takes variable from parent
print(d.ram)                    # Takes variable from Dell

d.laptop_info()                 # Takes method from parent
d.dell_info()                   # Takes method from Dell


# ==========================================
# 5. HYBRID INHERITANCE
# Combination of two or more types
# ==========================================

class Device:
    battery = 100

    def device_info(self):
        print("This is a device.")


class Laptop5(Device):          # Device → Laptop5
    ram = 16

    def laptop_info(self):
        print("This is a laptop.")


class Mobile(Device):           # Device → Mobile
    camera = 48

    def mobile_info(self):
        print("This is a mobile.")


class SmartDevice(Laptop5, Mobile):   # Multiple inheritance
    storage = 512

    def smart_info(self):
        print("This is a smart device.")


s = SmartDevice()

print("\n5. HYBRID INHERITANCE")
print(s.battery)                # Takes variable from Device
print(s.ram)                    # Takes variable from Laptop5
print(s.camera)                 # Takes variable from Mobile
print(s.storage)                # Takes variable from SmartDevice

s.device_info()                 # Takes method from Device
s.laptop_info()                 # Takes method from Laptop5
s.mobile_info()                 # Takes method from Mobile
s.smart_info()                  # Takes method from SmartDevice
```

```python
#Hierarchical Inheritance example
class employee:
  def __init__(self,name,emp_id,salary):
    self.name=name
    self.emp_id=emp_id
    self.salary=salary
  def display(self):
    print("Name: ",self.name)
    print("Employee ID: ",self.emp_id)
    print("Salary: ",self.salary)

class teacher(employee):
  def __init__(self,name,emp_id,salary,subject):
    super().__init__(name,emp_id,salary)
    self.subject=subject
  def display(self):
    super().display()
    print("Subject: ",self.subject)

class adminstaff(employee):
  def __init__(self,name,emp_id,salary,department):
    super().__init__(name,emp_id,salary)
    self.department=department
  def display(self):
    super().display()
    print("Department: ",self.department)

emp1=teacher("John",101,50000,"Maths")
emp2=adminstaff("Alice",102,60000,"HR")
emp1.display()
emp2.display()
```
