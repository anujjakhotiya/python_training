```python
import threading

buffer = []
BUFFER_SIZE = 5
mutex = threading.Lock()

def producer():
    global buffer
    mutex.acquire()
    print("Producer LOCKED")
    for i in range(BUFFER_SIZE):
        buffer.append(i)
        print("Produced:", i, "| Buffer:", buffer)
    print("Producer UNLOCKED")
    mutex.release()

def consumer():
    global buffer
    mutex.acquire()
    print("Consumer LOCKED")
    while len(buffer) > 0:
        item = buffer.pop(0)
        print("Consumed:", item, "| Buffer:", buffer)
    print("Consumer UNLOCKED")
    mutex.release()

p = threading.Thread(target=producer)
c = threading.Thread(target=consumer)
p.start()
p.join()
c.start()
c.join()
print("Done! Buffer:", buffer)
```
