```python
l=[4,2,6,3,7]
stack=[]
l2=[]
s=len(l)
for i in range(len(l)):
  while stack and l[i]>l[stack[-1]]:
    idx=stack.pop()
    l2.insert(idx,l[i])
  stack.append(i)
while len(stack)!=0:
  idx=stack.pop()
  l2.insert(idx,-1)
print(l2)
```

```python
class Solution:
    def finalPrices(self, prices):
        ans = prices[:]
        stack = []

        for i in range(len(prices)):
            while stack and prices[stack[-1]] >= prices[i]:
                j = stack.pop()
                ans[j] -= prices[i]
            stack.append(i)
        return ans
```
