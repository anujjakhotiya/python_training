```python
class Node:
    def __init__(self, data, nxt=None):
        self.data = data
        self.nxt = nxt


def insert_begin(head, val):
    new_node = Node(val)
    new_node.nxt = head
    return new_node


def insert_end(head, val):
    new_node = Node(val)

    if head is None:
        return new_node

    curr = head
    while curr.nxt:
        curr = curr.nxt

    curr.nxt = new_node
    return head


def reverse_list(head):
    prev = None
    curr = head

    while curr:
        next_node = curr.nxt
        curr.nxt = prev
        prev = curr
        curr = next_node

    return prev


def middle(head):
    slow = head
    fast = head

    while fast and fast.nxt:
        slow = slow.nxt
        fast = fast.nxt.nxt

    return slow


def display(head):
    curr = head

    while curr:
        print(curr.data, end=" -> ")
        curr = curr.nxt

    print("None")


# Driver
head = None

head = insert_end(head, 10)
head = insert_end(head, 20)
head = insert_end(head, 30)
head = insert_begin(head, 5)

display(head)

print("Middle:", middle(head).data)

head = reverse_list(head)

display(head)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.nxt = None


def insert_begin(head, val):
    new_node = Node(val)

    if head:
        head.prev = new_node
        new_node.nxt = head

    return new_node


def insert_end(head, val):
    new_node = Node(val)

    if head is None:
        return new_node

    curr = head

    while curr.nxt:
        curr = curr.nxt

    curr.nxt = new_node
    new_node.prev = curr

    return head


def display_forward(head):
    curr = head

    while curr:
        print(curr.data, end=" <-> ")

        if curr.nxt is None:
            tail = curr

        curr = curr.nxt

    print("None")

    return tail


def display_backward(tail):
    curr = tail

    while curr:
        print(curr.data, end=" <-> ")
        curr = curr.prev

    print("None")


# Driver
head = None

head = insert_end(head, 10)
head = insert_end(head, 20)
head = insert_end(head, 30)
head = insert_begin(head, 5)

tail = display_forward(head)

display_backward(tail)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.nxt = None


def insert_end(head, val):
    new_node = Node(val)

    if head is None:
        new_node.nxt = new_node
        return new_node

    curr = head

    while curr.nxt != head:
        curr = curr.nxt

    curr.nxt = new_node
    new_node.nxt = head

    return head


def insert_begin(head, val):
    new_node = Node(val)

    if head is None:
        new_node.nxt = new_node
        return new_node

    curr = head

    while curr.nxt != head:
        curr = curr.nxt

    curr.nxt = new_node
    new_node.nxt = head

    return new_node


def display(head):
    if head is None:
        print("Empty")
        return

    curr = head

    while True:
        print(curr.data, end=" -> ")
        curr = curr.nxt

        if curr == head:
            break

    print("(back to head)")


# Driver
head = None

head = insert_end(head, 10)
head = insert_end(head, 20)
head = insert_end(head, 30)
head = insert_begin(head, 5)

display(head)
```

```python
nums = [10,20,30,40,50]
val = int(input("enter a integer:"))

print(nums[2])
def linear_search(nums,val):

    for i in range(len(nums)):
        if nums[i] == val:
            return True,i

    return False
print(linear_search(nums,val))

def binary_search(nums,val):
    low = 0
    high = len(nums)-1

    while low <= high:

        mid = (low+high)//2

        if nums[mid] == val:
            return True,mid
            break

        elif nums[mid] < val:
            low = mid + 1
        else:
            high = mid - 1


print(binary_search(nums,val))
```

```python
def bubble_sort(nums):
    n = len(nums)
    for i in range(n-1):
        for j in range(n-1-i):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1] , nums[j]

    return nums
def selection_sort(nums):
    n = len(nums)
    for i in range(n - 1):

        min_index = i

        for k in range(i + 1,n):

            if nums[k] < nums[min_index]:
                min_index = k

            nums[i], nums[min_index] = nums[min_index], nums[i]
    return nums

def insertion_sort(nums):
    n = len(nums)
    for i in range(1,n):
        keys = nums[ i + 1]
        j = i -1
        while j >= 0 and nums[j] > keys:
            nums[j + 1] = nums[j]
            j -= 1

            nums[j+1] = keys
        return nums

def merge_sort(nums):
    if len(nums) <= 1:
        return nums
    mid = len(nums) // 2

    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    i = 0
    j = 0
    merged = []
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    while i < len(left):
        merged.append(left[i])
        i += 1

    while j < len(right):
        merged.append(right[j])
        j += 1
    return merged


nums = [1,5,2,7,99,0]

print(bubble_sort(nums))
print(selection_sort(nums))
print(insertion_sort(nums))
print(merge_sort(nums))

```
