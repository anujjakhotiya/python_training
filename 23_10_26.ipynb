```python
stack = []
n = int(input("Enter number of elements: "))

for i in range(n):
    x = int(input("Enter element: "))
    stack.append(x)       # Push

print("Stack:", stack)

# Push
x = int(input("Enter element to push: "))
stack.append(x)
print("After push:", stack)

# Pop
if len(stack) == 0:
    print("Stack Underflow")
else:
    print("Popped element:", stack.pop())

# Peek
if len(stack) == 0:
    print("Stack is empty")
else:
    print("Top element:", stack[-1])

# Display
print("Stack:", stack)
```

```python

class MyStack:
    def __init__(self):
        self.arr = []

    def push(self, x):
        self.arr.append(x)

    def pop(self):
        if len(self.arr) == 0:
            return -1
        return self.arr.pop()

    def peek(self):
        if len(self.arr) == 0:
            return -1
        return self.arr[-1]

    def getSize(self):
        return len(self.arr)

    def isEmpty(self):
        return len(self.arr) == 0
```

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class MyStack:
    def __init__(self):
        self.top = None
        self.size = 0

    def push(self, x):
        new_node = Node(x)
        new_node.next = self.top
        self.top = new_node
        self.size += 1

    def pop(self):
        if self.top is None:
            return -1

        x = self.top.data
        self.top = self.top.next
        self.size -= 1
        return x

    def peek(self):
        if self.top is None:
            return -1
        return self.top.data

    def getSize(self):
        return self.size

    def isEmpty(self):
        return self.top is None
```
