```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Playlist:
    def __init__(self):
        self.head = None

    # 1. Add song at beginning
    def add_begin(self, song):
        nn = Node(song)

        nn.next = self.head
        self.head = nn

    # 2. Add song at end
    def add_end(self, song):
        nn = Node(song)

        if self.head is None:
            self.head = nn
            return

        temp = self.head

        while temp.next is not None:
            temp = temp.next

        temp.next = nn

    # 3. Insert song at specified position
    def insert_pos(self, pos, song):
        if pos <= 0:
            print("Invalid position")
            return

        if pos == 1:
            self.add_begin(song)
            return

        temp = self.head
        i = 1

        while i < pos - 1 and temp is not None:
            temp = temp.next
            i += 1

        if temp is None:
            print("Invalid position")
            return

        nn = Node(song)
        nn.next = temp.next
        temp.next = nn

    # 4. Remove first song
    def delete_begin(self):
        if self.head is None:
            print("Playlist is empty")
            return

        self.head = self.head.next

    # 5. Remove last song
    def delete_end(self):
        if self.head is None:
            print("Playlist is empty")
            return

        if self.head.next is None:
            self.head = None
            return

        temp = self.head

        while temp.next.next is not None:
            temp = temp.next

        temp.next = None

    # 6. Remove song at specified position
    def delete_pos(self, pos):
        if self.head is None:
            print("Playlist is empty")
            return

        if pos <= 0:
            print("Invalid position")
            return

        if pos == 1:
            self.delete_begin()
            return

        temp = self.head
        i = 1

        while i < pos - 1 and temp is not None:
            temp = temp.next
            i += 1

        if temp is None or temp.next is None:
            print("Invalid position")
            return

        temp.next = temp.next.next

    # 7. Search for a song
    def search(self, song):
        temp = self.head
        pos = 1

        while temp is not None:
            if temp.data == song:
                return pos

            temp = temp.next
            pos += 1

        return -1

    # 8. Count songs
    def count(self):
        temp = self.head
        c = 0

        while temp is not None:
            c += 1
            temp = temp.next

        return c

    # 9. Display playlist
    def display(self):
        if self.head is None:
            print("Playlist is empty")
            return

        temp = self.head

        while temp is not None:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")

    # 10. Reverse playlist
    def reverse(self):
        prev = None
        temp = self.head

        while temp is not None:
            next_node = temp.next
            temp.next = prev
            prev = temp
            temp = next_node

        self.head = prev


# Main Program
pl = Playlist()

while True:

    print("\n----- MUSIC PLAYLIST MANAGER -----")
    print("1. Add song at beginning")
    print("2. Add song at end")
    print("3. Insert song at position")
    print("4. Remove first song")
    print("5. Remove last song")
    print("6. Remove song at position")
    print("7. Search for a song")
    print("8. Count songs")
    print("9. Display playlist")
    print("10. Reverse playlist")
    print("11. Exit")

    choice = int(input("Enter your choice: "))

    match choice:

        case 1:
            song = input("Enter song name: ")
            pl.add_begin(song)

        case 2:
            song = input("Enter song name: ")
            pl.add_end(song)

        case 3:
            pos = int(input("Enter position: "))
            song = input("Enter song name: ")
            pl.insert_pos(pos, song)

        case 4:
            pl.delete_begin()

        case 5:
            pl.delete_end()

        case 6:
            pos = int(input("Enter position: "))
            pl.delete_pos(pos)

        case 7:
            song = input("Enter song to search: ")
            pos = pl.search(song)

            if pos == -1:
                print("Song not found")
            else:
                print("Song found at position:", pos)

        case 8:
            print("Total number of songs:", pl.count())

        case 9:
            pl.display()

        case 10:
            pl.reverse()
            print("Playlist reversed")

        case 11:
            print("Exiting application...")
            break

        case _:
            print("Invalid choice")
```
