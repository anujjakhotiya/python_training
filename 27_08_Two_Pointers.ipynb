```python
arr=[1,2,3,4,5]
left=0
right=len(arr)-1

while left<=right:
  sum=arr[left]+arr[right]
  print(sum)

  left+=1
  right-=1
```

```python
#reverse array using 2 pointer
arr=list(map(int,input().split()))
left=0
right=len(arr)-1

while left<=right:
  arr[left],arr[right]=arr[right],arr[left]
  left+=1
  right-=1
print(arr)
```

```python
l1=[-5,-2,3,4,6]
l2=[-6,-1,0,99]
l3=[]
l1.sort()
l2.sort()
left=0
right=0
while left<len(l1) and right<len(l1):
  if l1[left]<l2[right]:
    l3.append(l1[left])
    left+=1
  else:
    l3.append(l2[right])
    right+=1

while left<len(l1):
    l3.append(l1[left])
    left+=1
while right<len(l2):
    l3.append(l2[right])
    right+=1

print(l3)
```

```python
list1=list(map(int,input().split()))
list2=list(map(int,input().split()))
i=0
j=0
result=[]

while i<len(list1) and j<len(list2):
    if list1[i]<list2[j]:
        result.append(list1[i])
        i+=1
    else:
        result.append(list2[j])
        j+=1
while i<len(list1):
    result.append(list1[i])
    i+=1
while j<len(list2):
    result.append(list2[j])
    j+=1
print(result)
```

```python
a=str(input())
left=0
right=len(a)-1
flag=0
while(left<right):
  if a[left]!=a[right]:
    flag=1
    break
  else:
    left+=1
    right-=1
if flag==1:
  print("NO")
else:
  print("YES")
```

Sliding Window:


```python
l=[23,36,42,39,88,24,60,12,19]
k=4
start=0
end=k-1
max=0
while end<= len(l)-1:
  l2=l[start:end]
  s=sum(l2)
  if s>max:
    max=s
  start+=1
  end+=1
print(max)
```
