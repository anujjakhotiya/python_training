Bubble sort:

```python
n=int(input())
arr =list(map(int,input().split()))
for i in range(n):
  for j in range(i+1, n):
    if (arr[i]>arr[j]):
      arr[i],arr[j]=arr[j],arr[i]
print(arr)
```

```python
n=int(input())
arr =list(map(int,input().split()))
for i in range(n-1):
  for j in range(n-i-1):
    if (arr[j]>arr[j+1]):
      arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

Selection sort:
->you r gng to select a value and swap it
->if you find min->swap with first
->if you find max->swap with last

```python
n=int(input())
arr =list(map(int,input().split()))
for i in range(n):
  idx=i
  for j in range(i+1, n):
    if (arr[j]<arr[idx]):
      idx=j
  arr[i],arr[idx]=arr[idx],arr[i]
print(arr)
```

Insertion sort:
->get one value from array
->

```python
n=int(input())
arr =list(map(int,input().split()))
for i in range(1,n):
  val=arr[i]
  j=i-1
  while(j>=0 and arr[j]>val):
    arr[j+1]=arr[j]
    j=j-1
  arr[j+1]=val
print(arr)
```

```python
'''
  .**...**.
  ****.****
  *********
  .*******
  ..*****
  ...***
  ....*
  print the Pattern
'''
n = int(input("Enter n: "))

# Top part
for i in range(2):
    stars = 2 * (i + 1)
    dots = n - 4 - 2 * i
    print("." + "*" * stars + "." * dots + "*" * stars)

# Middle row
print("*" * (n + 2))

# Bottom part
for i in range((n + 1) // 2):
    dots = i + 1
    stars = n - 2 * i
    print("." * dots + "*" * stars)
```
